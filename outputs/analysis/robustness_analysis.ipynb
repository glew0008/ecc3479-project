# Robustness Analysis of Education-Income Association

This notebook evaluates the main descriptive finding from the primary econometric analysis: higher education rank is associated with higher estimated weekly income for Australian males, using ABS Census data from 2016 and 2021.

The robustness checks include: alternative control sets, alternative sample definitions, an alternative functional form, and alternative inference. The results are presented in a side-by-side table with sample size and notes for each variation.

```python
from pathlib import Path
import sys

project_root = Path().resolve()
for candidate in [project_root, *project_root.parents]:
    if (candidate / 'src').exists() and (candidate / 'data' / 'clean').exists():
        project_root = candidate
        break
else:
    raise FileNotFoundError('Could not find project root containing src/ and data/clean/')

sys.path.insert(0, str(project_root / 'src'))

import pandas as pd
from robustness_analysis import load_clean_data, build_robustness_table

clean_path = project_root / 'data' / 'clean' / 'merged_2016_2021.csv'
df = load_clean_data(clean_path)
summary_table = build_robustness_table(df)
summary_table
```

```python
output_path = project_root / 'outputs' / 'analysis' / 'robustness_table.csv'
summary_table.to_csv(output_path)
print(f'Results saved to: {output_path}')
```

## Interpretation

Each specification is designed to test a different potential fragility in the descriptive association between education rank and estimated weekly income.

- If the education_rank coefficient remains positive and statistically meaningful across the checks, the main descriptive finding is robust to sample composition, functional-form assumptions, and standard error specification.
- If estimates change materially when the top income bracket is excluded or when only 2021 is used, that would point to sensitivity driven by extreme values or by changing year composition.

The saved CSV file also provides a reproducible table for reporting and comparison.
